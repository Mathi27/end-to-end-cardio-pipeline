```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import os
from google.colab import files

# --- 1. LOAD DATA ---
INPUT_PATH = '/content/data/02_processed/heart_disease_combined.csv'
REPORTS_DIR = '/content/reports/figures'
os.makedirs(REPORTS_DIR, exist_ok=True)

if not os.path.exists(INPUT_PATH):
    print(f"❌ Error: {INPUT_PATH} not found. Please upload it or run Notebook 01.")
else:
    print(f"📂 Loading data from: {INPUT_PATH}")
    df = pd.read_csv(INPUT_PATH)

    # --- 2. GENERATE CHARTS ---
    print("\n📊 Generating Charts...")

    # Chart 1: Target Distribution
    plt.figure(figsize=(6, 4))
    sns.countplot(x='target', data=df, hue='target', legend=False, palette='coolwarm')
    plt.title("Heart Disease Distribution (0=Healthy, 1=Disease)")
    dist_path = os.path.join(REPORTS_DIR, 'target_distribution.png')
    plt.savefig(dist_path)
    plt.close()
    
    # Chart 2: Correlation Heatmap
    plt.figure(figsize=(10, 8))
    numeric_cols = ['age', 'trestbps', 'chol', 'thalach', 'oldpeak', 'target']
    sns.heatmap(df[numeric_cols].corr(), annot=True, cmap='RdBu_r', fmt=".2f")
    plt.title("Feature Correlation")
    heatmap_path = os.path.join(REPORTS_DIR, 'correlation_heatmap.png')
    plt.savefig(heatmap_path)
    plt.close()

    print(f"✅ Charts saved to: {REPORTS_DIR}")

    # --- 3. DOWNLOAD ---
    print("⬇️  Downloading charts...")
    files.download(dist_path)
    files.download(heatmap_path)
```
